## 🔹 Ενότητα 2 — Επεξήγηση κώδικα - Η λειτουργία των μοτέρ του 3D scanner

Κάθε κινητήρας για να λειτουργήσει θα πρέπει να ελέγχεται από έναν οδηγό (motor driver). Στην περίπτωση μας οι κινητήρες Nema 17 που χρησιμοποιούμε έχουν ως motor drivers τους **A4988**. Ο οδηγός στέλνει ηλεκτρικούς παλμούς στο μοτέρ και εκείνο κάνει μικρά βηματάκια (STEPs). Τα μοτέρ που χρησιμοποιούμε εδώ για να ολοκληρώσουν μια πλήρη περιστροφή 360 μοιρών του άξονα τους πρέπει να κάνουν 200 βήματα.

Μικρή άσκηση: Αν για να γίνει μια περιστροφή του μοτερ 360 μοιρών χρειάζονται 200 steps, τότε το 1 step πόσες μοίρες περιστρέφει τον άξονα;

Ο οδηγός για να οδηγήσει το μοτέρ χρειάζεται 3 σήματα:

| Σήμα | Τι σημαίνει |
|------|-------------|
| STEP | Κάνε ένα μικρό βήμα |
| DIR  | Προς ποια κατεύθυνση |
| EN   | Ενεργοποίηση ή απενεργοποίηση του μοτερ|

### ✏️ Μικρή Άσκηση
Αν ένα μοτέρ έχει **1.8° ανά βήμα**, πόσες μοίρες θα γυρίσει με 10 βήματα;


Πάμε όμως να δούμε το σύνολο του κώδικα που μας έφτιαξε το ΑΙ ποιο αναλυτικά και να μελετήσουμε τη δομή του. Το ESP32 προγραμματίζεται στο περιβάλλον IDE που προγραμματίζουμε το Arduino και με τον ίδιο τρόπο και γλώσσα προγραμματισμού.

Αρχικά λοιπόν κάνουμε συμπερίληψη των απαραίτητων βιβλιοθηκών

```
#include <WiFi.h>
#include <WebServer.h>
```
αυτές οι βιβλιοθήκες δίνουν τη δυναότητα στο EPS32 να δημιουργήσει το Wi-Fi και να συνδέεται και να επικοινωνεί με συσκευές μέσω ιστοσελίδας

στη συνέχεια ορίζουμε το τι κάνουν τα pin του ESP32 στα οποία έχουμε συνδέσει τα δύο μοτερ. Έχουμε 2 μοτερ  
```
// ================= PIN MAP =================
#define TILT_DIR   6
#define TILT_STEP  7
#define YAW_DIR    5
#define YAW_STEP   4
#define YAW_EN     20
#define TILT_EN    21
```




````
#include <WiFi.h>
#include <WebServer.h>

// ================= PIN MAP =================
#define TILT_DIR   6
#define TILT_STEP  7
#define YAW_DIR    5
#define YAW_STEP   4
#define YAW_EN     20
#define TILT_EN    21

// ================= PARAMETERS =================
const int yawStepsPerMove = 11;      // ≈20°
const int yawPositions    = 18;

const int tiltSteps_deg  = 32;     // calibration reference
const int settleDelay     = 500;    // ms
const int cameraInterval  = 2000;   // ms

// ================= STATE =================
bool scanning = false;
bool paused   = false;

int currentPhoto = 0;
const int totalPhotos = yawPositions * 3;

int yawIndex = 0;
float currentTiltDeg = 0;          // real physical tilt
int tiltDegSetting = 40;           // default tilt angle

WebServer server(80);

// ================= MOTOR CONTROL =================
void stepMotor(int stepPin, int dirPin, int steps, bool dir) {
  digitalWrite(YAW_EN, LOW);
  digitalWrite(TILT_EN, LOW);

  digitalWrite(dirPin, dir ? HIGH : LOW);
  for (int i = 0; i < steps; i++) {
    digitalWrite(stepPin, HIGH);
    delayMicroseconds(1200);
    digitalWrite(stepPin, LOW);
    delayMicroseconds(1200);
  }
}

int tiltStepsForDeg(float deg) {
  return round(deg * (tiltSteps_deg / 15.0));
}

void moveTiltTo(float targetDeg) {
  float delta = targetDeg - currentTiltDeg;
  if (abs(delta) < 0.5) return;

  bool dir = delta > 0;
  int steps = tiltStepsForDeg(abs(delta));

  stepMotor(TILT_STEP, TILT_DIR, steps, dir);
  currentTiltDeg = targetDeg;
}

// ================= PAUSE =================
void waitIfPaused() {
  while (paused) {
    server.handleClient();
    delay(50);
  }
}

// ================= SCAN =================
void scanAtCurrentTilt() {
  for (int i = 0; i < yawPositions; i++) {
    waitIfPaused();
    stepMotor(YAW_STEP, YAW_DIR, yawStepsPerMove, true);
    yawIndex = (yawIndex + 1) % yawPositions;
    delay(settleDelay);
    delay(cameraInterval);
    currentPhoto++;
  }
}

void fullScan() {
  scanning = true;
  paused   = false;
  currentPhoto = 0;
  yawIndex = 0;

  moveTiltTo(0);
  scanAtCurrentTilt();

  moveTiltTo(tiltDegSetting);
  delay(800);
  scanAtCurrentTilt();

  moveTiltTo(-tiltDegSetting);
  delay(800);
  scanAtCurrentTilt();

  moveTiltTo(0);

  scanning = false;

  digitalWrite(YAW_EN, HIGH);
  digitalWrite(TILT_EN, HIGH);
}

// ================= WEB UI =================
void handleRoot() {
  int percent = (currentPhoto * 100) / totalPhotos;
  int yawDeg  = yawIndex * 15;

  server.send(200, "text/html",
    "<h2>3D scanner – CDG, Cretan Dynamic Girls</h2>"

    "<p><b>Yaw:</b> " + String(yawDeg) + "° &nbsp; "
    "<b>Tilt:</b> " + String(currentTiltDeg, 1) + "°</p>"

    "<p><b>Photos:</b> " + String(currentPhoto) +
    " / " + String(totalPhotos) + "</p>"

    "<p><b>Progress:</b> " + String(percent) + "%</p>"

    "<hr>"

    "<p><b>Tilt angle (deg):</b></p>"
    "<input type='number' id='t' value='" + String(tiltDegSetting) + "'>"
    "<button onclick=\"fetch('/setTilt?d='+document.getElementById('t').value)\">Set</button><br><br>"

    "<button onclick=\"fetch('/start')\">Start Scan</button> "
    "<button onclick=\"fetch('/pause')\">Pause</button> "
    "<button onclick=\"fetch('/resume')\">Resume</button><br><br>"

    "<button onclick=\"fetch('/zero')\">Set Zero</button><br><br>"

    "<button onclick=\"fetch('/yaw')\">Yaw +15°</button><br><br>"

    "<button onclick=\"fetch('/tiltUp')\">Tilt +</button> "
    "<button onclick=\"fetch('/tiltDown')\">Tilt -</button><br><br>"

    "<button onclick=\"location.reload()\">Refresh</button>"
  );
}

// ================= SETUP =================
void setup() {
  pinMode(TILT_DIR, OUTPUT);
  pinMode(TILT_STEP, OUTPUT);
  pinMode(YAW_DIR, OUTPUT);
  pinMode(YAW_STEP, OUTPUT);
  pinMode(YAW_EN, OUTPUT);
  pinMode(TILT_EN, OUTPUT);

  digitalWrite(YAW_EN, LOW);
  digitalWrite(TILT_EN, LOW);

  WiFi.softAP("3D-Scanner", "scanner123");

  server.on("/", handleRoot);

  server.on("/start", [](){ if (!scanning) fullScan(); server.send(200,"text/plain","Scanning"); });
  server.on("/pause", [](){ paused = true; server.send(200,"text/plain","Paused"); });
  server.on("/resume", [](){ paused = false; server.send(200,"text/plain","Resumed"); });

  server.on("/zero", [](){
    yawIndex = 0;
    currentTiltDeg = 0;
    server.send(200,"text/plain","Zero set");
  });

  server.on("/yaw", [](){
    stepMotor(YAW_STEP, YAW_DIR, yawStepsPerMove, true);
    yawIndex = (yawIndex + 1) % yawPositions;
    server.send(200,"text/plain","Yaw moved");
  });

  server.on("/tiltUp", [](){
    moveTiltTo(tiltDegSetting);
    server.send(200,"text/plain","Tilt up");
  });

  server.on("/tiltDown", [](){
    moveTiltTo(-tiltDegSetting);
    server.send(200,"text/plain","Tilt down");
  });

  server.on("/setTilt", [](){
    if (server.hasArg("d")) {
      tiltDegSetting = constrain(server.arg("d").toInt(), 5, 45);
    }
    server.send(200,"text/plain","Tilt angle set");
  });

  server.begin();
}

void loop() {
  server.handleClient();
}
````
