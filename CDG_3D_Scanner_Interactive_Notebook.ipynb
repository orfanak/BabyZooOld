## 🔹 Ενότητα 2 — Επεξήγηση κώδικα - Η λειτουργία των μοτέρ του 3D scanner

Κάθε κινητήρας για να λειτουργήσει θα πρέπει να ελέγχεται από έναν οδηγό (motor driver). Στην περίπτωση μας οι κινητήρες Nema 17 που χρησιμοποιούμε έχουν ως motor drivers τους **A4988**. Ο οδηγός στέλνει ηλεκτρικούς παλμούς στο μοτέρ και εκείνο κάνει μικρά βηματάκια (STEPs). Τα μοτέρ που χρησιμοποιούμε εδώ για να ολοκληρώσουν μια πλήρη περιστροφή 360 μοιρών του άξονα τους πρέπει να κάνουν 200 βήματα.

Μικρή άσκηση: Αν για να γίνει μια περιστροφή του μοτερ 360 μοιρών χρειάζονται 200 steps, τότε το 1 step πόσες μοίρες περιστρέφει τον άξονα;




## 🔹 Ενότητα 3 — Πώς κινείται ένα μοτέρ με παλμούς (pulses);

Ο οδηγός A4988 «καταλαβαίνει» κίνηση μόνο όταν του στείλουμε έναν παλμό STEP.

Ένας παλμός είναι:

```cpp
digitalWrite(STEP_PIN, HIGH); // δώσε τάση στο Pin του κινητήρα για να κάνεις ένα step
delayMicroseconds(1200); // περίμενε 1200 εκατομυριοστά του δευτερολέπτου έχοντας την τάση σε HIGH
digitalWrite(STEP_PIN, LOW); // μηδένισε την τάση (LOW) για να τελείωσει η κίνηση του μοτερ



### ✏️ Μικρή Άσκηση
Αν ένα μοτέρ έχει **1.8° ανά βήμα**, πόσες μοίρες θα γυρίσει με 10 βήματα;


Πάμε όμως να δούμε το σύνολο του κώδικα που μας έφτιαξε το ΑΙ ποιο αναλυτικά και να μελετήσουμε τη δομή του. Το ESP32 προγραμματίζεται στο περιβάλλον IDE που προγραμματίζουμε το Arduino και με τον ίδιο τρόπο και γλώσσα προγραμματισμού. Όλος ο κώδικας είναι διαθέσιμος στο τέλος αυτού του Notebook.

Αρχικά λοιπόν κάνουμε συμπερίληψη των απαραίτητων βιβλιοθηκών

```

// ================= PIN MAP =================
#define TILT_DIR   6
#define TILT_STEP  7
#define YAW_DIR    5
#define YAW_STEP   4
#define YAW_EN     20
#define TILT_EN    21
```

Ο οδηγός για να οδηγήσει το μοτέρ χρειάζεται 3 σήματα:

| Σήμα | Τι σημαίνει |
|------|-------------|
| STEP | Κάνε ένα μικρό βήμα |
| DIR  | Προς ποια κατεύθυνση |
| EN   | Ενεργοποίηση ή απενεργοποίηση του μοτερ|



## 🔹 Ενότητα 4 — Μεταβλητές που ρυθμίζουν τον scanner

Αυτές οι μεταβλητές καθορίζουν πώς κινείται ο scanner:

| Μεταβλητή | Ρόλος |
|-----------|------|
| yawStepsPerMove | Πόσα steps κάνει το μοτέρ YAW κάθε φορά |
| yawPositions | Πόσες στάσεις (για αντίστοιχες φωτογραφίες) για έναν πλήρη κύκλο |
| tiltSteps | Πόσα steps κάνει το TILT μοτέρ |
| cameraInterval | Πόσο περιμένουμε μεταξύ φωτογραφιών (ms) |

📌 Παράδειγμα από τον κώδικα:

```cpp
const int yawStepsPerMove = 11; //11 steps αντιστοιχούν σε περίπου 20 μοίρες περιστροφης 11*1.8ο μοιρες = 19.8ο μοίρες
const int yawPositions    = 18; // 18 φωτογραφίες για μια πλήρη περιστροφή (360ο μοιρες ο κυκλος / 20ο μοιρες καθε σταστη = 18 φωτογραφιες
const int tiltSteps       = 80; //τα 80 steps στο TILT δίνουν καλές γωνίες φωτογράφισης πανω και κάτω από το οριζόντιο επιπεδο αντιστοιχα
const int cameraInterval = 2000; // χρονος αναμονης αναμεσα στις φωτογραφιες και τα βηματα, περιμένουμε 2 δευτερόλεπτα για κάθε φωτογραφία




## 🔹 Ενότητα 5 — Η συνάρτηση stepMotor()

Αυτή είναι η βασική συνάρτηση κίνησης:

```cpp
void stepMotor(int stepPin, int dirPin, int steps, bool dir)
```

Ενεργοποιεί το μοτέρ που του προσδιορίζουμε στις μεταβλητες και στέλνει παλμούς.

Αυτή η συνάρτηση:

• επιλέγει ποιο μοτέρ θα κινηθεί μεσω του καταλληλου Pin, int stepPin
• ορίζει την κατεύθυνση περιστροφής του μοτερ, int dirPin
• στέλνει όσα steps χρειάζονται να γινουν για καθε σταση, int steps







## 🔹 Ενότητα 6 — Κίνηση σε συγκεκριμένη γωνία (Tilt)


Στον scanner μας για το TILT **δεν μετράμε μοίρες**.

Μετράμε:
> πόσα steps χρειάζεται το TILT μοτέρ για να παει πιο ψηλά και πιο χαμηλά από το οριζόντιο επίπεδο
> Η τιμή που βάζουμε στο TILT είναι σε steps οχι σε μοίρες και έχει προκύψει από παρατήρηση ώστε να έχουμε μια καλή γωνία πάνω και κάτω από το όριζόντιο επίπεδο για να πάρουμε καλές φωτογραφίες.

Από πειραματισμό παρατηρήσαμε ότι:
≈ 80 steps δίνουν καλή κλίση για scanning.
Οι παρακάτω συναρτήσεις χρησιμοποιούνται για να ανεβάσουν ή να κατεβάσουν το TILT

```
void moveTiltUp() {
  stepMotor(TILT_STEP, TILT_DIR, tiltSteps, true);
}

void moveTiltDown() {
  stepMotor(TILT_STEP, TILT_DIR, tiltSteps, false);
}

```





## 🔹 Ενότητα 7 — Πλήρης σάρωση του αντικειμένου

Η πλήρης σάρωση γίνεται ως εξής:

1️⃣ Το αντικείμενο περιστρέφεται στο κέντρο (YAW)  
με τον εξής κώδικά για ένα πλήρη κύκλο με βάση τα βήματα και τις στάσεις που έχουμε επιλέξει στις αρχικές μεταβλητές μας μέσω της παρακάτω συνάρτησης:
```
void scanOnce() {
  for (int i = 0; i < yawPositions; i++) {
    stepMotor(YAW_STEP, YAW_DIR, yawStepsPerMove, true);
    delay(500);
    delay(2000);
  }
}

Για μια πλήρη σάρωση του αντικειμένου που περιλαμβάνει και τα διαφορετικά TILT, η παραπάνω συνάρτηση τρέχει 3 φορές
1ο για μια σάρωση στο οριζόντιο επίπεδο (ένας πλήρης YAW κύκλος)
και αφου ολοκληρωθεί η αρχική αυτή σάρωση στο οριζόντιο επίπεδο προχωράμε με τη σάρωση σε διαφορετικά TILT:

2️⃣ Το TILT μοτέρ κινείται προς τα πάνω (80 steps) και ξαναγίνεται περιστροφή YAW

3️⃣ Το TILT μοτέρ κινείται προς τα κάτω (- 2 × 80 steps) και ξαναγίνεται περιστροφή YAW

4️⃣ Το σύστημα επιστρέφει στο κέντρο και σταματά η διαδικασία

📌 Όλη η διαδικασία γίνεται **αυτόματα**.
Ο χρήστης απλώς πατάει **Start Scan**.

Ο κώδικας της πλήρους σάρωσης που περιγράφεται παραπάνω είναι ο εξής:

void fullScan() {

  // --- Tilt center ---
  scanOnce();

  // --- Tilt UP ---
  moveTiltUp();
  delay(800);
  scanOnce();

  // --- Tilt DOWN ---
  moveTiltDown();
  moveTiltDown();   // go below center
  delay(800);
  scanOnce();

  // --- Back to center ---
  moveTiltUp();

  // Disable motors (cool down)
  digitalWrite(YAW_EN, HIGH);
  digitalWrite(TILT_EN, HIGH);
}



## 🔹 Ενότητα 8 — Το Wi-Fi Control Panel

Το ESP32 με τις παρακάτω εντολές δημιουργεί wifi server "3D-scanner" password: "scanner123"

```

```

Για να χρησιμοποιήσουμε το scanner συνδεόμαστε από μια συσκεύη στο wifi αυτό και ανοίγουμε web σελίδα:
192.168.4.1

Η μορφή της ιστοσελίδας καθορίζεται από τον παρακάτω html κώδικα τον οποίο μπορούμε εύκολα να τροποποιήσουμε:


// ================= WEB UI =================

void handleRoot() {
  server.send(200, "text/html",
    "<h2>3D scanner – CDG, Cretan Dynamic Girls</h2>"

    "<p><b>Yaw:</b> 18 photos / full rotation</p>"
    "<p><b>Tilt:</b> ±" + String(tiltSteps) + " steps</p><hr>"

    "<button onclick=\"fetch('/start')\">Start Scan</button><br><br>"
    "<button onclick=\"fetch('/yaw')\">Yaw Step</button><br><br>"
    "<button onclick=\"fetch('/tiltUp')\">Tilt Up</button> "
    "<button onclick=\"fetch('/tiltDown')\">Tilt Down</button>"
  );
}

Οι γραμμές κώδικα μέσα στις αγκύλες είναι γραμμές σε γλώσσα html που χρησιμοποιείται για να φτιάχνοται ιστοσελίδες και με αυτές τις γραμμες κώδικα δημιουργούνται τα διάφορα κουμπιά που υπάρχουν στην ιστοσελίδα που ελέγχει το C3DG scanner

Τέλος υπάρχουν και κομμάτια πτου κώδικα που κάνουν το SETUP τα pins και διαχειρίζονται τον wifi server καθώς και η συνάρτηση void loop() που υπάρχει και στον προγραμματισμό με Arduino

// ================= SETUP =================


void setup() {
  pinMode(TILT_DIR, OUTPUT);
  pinMode(TILT_STEP, OUTPUT);
  pinMode(YAW_DIR, OUTPUT);
  pinMode(YAW_STEP, OUTPUT);
  pinMode(YAW_EN, OUTPUT);
  pinMode(TILT_EN, OUTPUT);

  digitalWrite(YAW_EN, LOW);
  digitalWrite(TILT_EN, LOW);

  WiFi.softAP("3D-Scanner", "scanner123");

  server.on("/", handleRoot);

  server.on("/start", [](){
    fullScan();
    server.send(200, "text/plain", "Scan started");
  });

  server.on("/yaw", [](){
    stepMotor(YAW_STEP, YAW_DIR, yawStepsPerMove, true);
    server.send(200, "text/plain", "Yaw moved");
  });

  server.on("/tiltUp", [](){
    moveTiltUp();
    server.send(200, "text/plain", "Tilt up");
  });

  server.on("/tiltDown", [](){
    moveTiltDown();
    server.send(200, "text/plain", "Tilt down");
  });

  server.begin();
}

void loop() {
  server.handleClient();
}




To σύνολο του κώδικα που χρησιμοποιούμε βρίσκεται εδώ:


```
#include <WiFi.h>
#include <WebServer.h>

// ================= PIN MAP =================
#define TILT_DIR   6
#define TILT_STEP  7
#define YAW_DIR    5
#define YAW_STEP   4
#define YAW_EN     20
#define TILT_EN    21


// ================= SCAN PARAMETERS =================

// Yaw: ~20° per move → 18 photos
const int yawStepsPerMove = 11;   // 1.8° × 11 ≈ 20°
const int yawPositions    = 18;

// Tilt: empirically chosen value
const int tiltSteps = 80;         // ⭐ good tilt for scanning

const int settleDelay    = 500;   // ms
const int cameraInterval = 2000;  // ms

WebServer server(80);

// ================= MOTOR CONTROL =================
void stepMotor(int stepPin, int dirPin, int steps, bool dir) {
  digitalWrite(YAW_EN, LOW);
  digitalWrite(TILT_EN, LOW);

  digitalWrite(dirPin, dir ? HIGH : LOW);

  for (int i = 0; i < steps; i++) {
    digitalWrite(stepPin, HIGH);
    delayMicroseconds(1200);
    digitalWrite(stepPin, LOW);
    delayMicroseconds(1200);
  }
}

void moveTiltUp() {
  stepMotor(TILT_STEP, TILT_DIR, tiltSteps, true);
}

void moveTiltDown() {
  stepMotor(TILT_STEP, TILT_DIR, tiltSteps, false);
}

// ================= SCAN =================
void scanOnce() {
  for (int i = 0; i < yawPositions; i++) {
    stepMotor(YAW_STEP, YAW_DIR, yawStepsPerMove, true);
    delay(settleDelay);
    delay(cameraInterval);
  }
}

void fullScan() {

  // --- Tilt center ---
  scanOnce();

  // --- Tilt UP ---
  moveTiltUp();
  delay(800);
  scanOnce();

  // --- Tilt DOWN ---

  scanOnce();

  // --- Back to center ---
  moveTiltUp();

  // Disable motors (cool down)
  digitalWrite(YAW_EN, HIGH);
  digitalWrite(TILT_EN, HIGH);
}

// ================= WEB UI =================

  );
}

// ================= SETUP =================
void setup() {
  pinMode(TILT_DIR, OUTPUT);
  pinMode(TILT_STEP, OUTPUT);
  pinMode(YAW_DIR, OUTPUT);
  pinMode(YAW_STEP, OUTPUT);
  pinMode(YAW_EN, OUTPUT);
  pinMode(TILT_EN, OUTPUT);

  digitalWrite(YAW_EN, LOW);
  digitalWrite(TILT_EN, LOW);

  WiFi.softAP("3D-Scanner", "scanner123");

  server.on("/", handleRoot);

  server.on("/start", [](){
    fullScan();
    server.send(200, "text/plain", "Scan started");
  });

  server.on("/yaw", [](){
    stepMotor(YAW_STEP, YAW_DIR, yawStepsPerMove, true);
    server.send(200, "text/plain", "Yaw moved");
  });

  server.on("/tiltUp", [](){
    moveTiltUp();
    server.send(200, "text/plain", "Tilt up");

  server.on("/tiltDown", [](){
