## 🔹 Ενότητα 2 — Επεξήγηση κώδικα - Η λειτουργία των μοτέρ του 3D scanner

Κάθε κινητήρας για να λειτουργήσει θα πρέπει να ελέγχεται από έναν οδηγό (motor driver). Στην περίπτωση μας οι κινητήρες Nema 17 που χρησιμοποιούμε έχουν ως motor drivers τους **A4988**. Ο οδηγός στέλνει ηλεκτρικούς παλμούς στο μοτέρ και εκείνο κάνει μικρά βηματάκια (STEPs). Τα μοτέρ που χρησιμοποιούμε εδώ για να ολοκληρώσουν μια πλήρη περιστροφή 360 μοιρών του άξονα τους πρέπει να κάνουν 200 βήματα.

Μικρή άσκηση: Αν για να γίνει μια περιστροφή του μοτερ 360 μοιρών χρειάζονται 200 steps, τότε το 1 step πόσες μοίρες περιστρέφει τον άξονα;

Ο οδηγός για να οδηγήσει το μοτέρ χρειάζεται 3 σήματα:

| Σήμα | Τι σημαίνει |
|------|-------------|
| STEP | Κάνε ένα μικρό βήμα |
| DIR  | Προς ποια κατεύθυνση |
| EN   | Ενεργοποίηση ή απενεργοποίηση του μοτερ|

```

Για να χρησιμοποιήσουμε το scanner συνδεόμαστε από μια συσκεύη στο wifi αυτό και ανοίγουμε web σελίδα:
192.168.4.1

Η μορφή της ιστοσελίδας καθορίζεται από τον παρακάτω html κώδικα τον οποίο μπορούμε εύκολα να τροποποιήσουμε:


// ================= WEB UI =================

void handleRoot() {
  server.send(200, "text/html",
    "<h2>3D scanner – CDG, Cretan Dynamic Girls</h2>"

    "<p><b>Yaw:</b> 18 photos / full rotation</p>"
    "<p><b>Tilt:</b> ±" + String(tiltSteps) + " steps</p><hr>"

    "<button onclick=\"fetch('/start')\">Start Scan</button><br><br>"
    "<button onclick=\"fetch('/yaw')\">Yaw Step</button><br><br>"
    "<button onclick=\"fetch('/tiltUp')\">Tilt Up</button> "
    "<button onclick=\"fetch('/tiltDown')\">Tilt Down</button>"
  );
}



και φυσικά υπάρχουν και κομμάτια πτου κώδικα που κάνουν το SETUP τα pins και διαχειρίζονται τον wifi server και η συνάρτηση void loop() όπως και στον προγραμματισμό με Arduino

// ================= SETUP =================


void setup() {
  pinMode(TILT_DIR, OUTPUT);
  pinMode(TILT_STEP, OUTPUT);
  pinMode(YAW_DIR, OUTPUT);
  pinMode(YAW_STEP, OUTPUT);
  pinMode(YAW_EN, OUTPUT);
  pinMode(TILT_EN, OUTPUT);

  digitalWrite(YAW_EN, LOW);
  digitalWrite(TILT_EN, LOW);

  WiFi.softAP("3D-Scanner", "scanner123");

  server.on("/", handleRoot);

  server.on("/start", [](){
    fullScan();
    server.send(200, "text/plain", "Scan started");
  });

  server.on("/yaw", [](){
    stepMotor(YAW_STEP, YAW_DIR, yawStepsPerMove, true);
    server.send(200, "text/plain", "Yaw moved");
  });

  server.on("/tiltUp", [](){
    moveTiltUp();
    server.send(200, "text/plain", "Tilt up");
  });

  server.on("/tiltDown", [](){
    moveTiltDown();
    server.send(200, "text/plain", "Tilt down");
  });

  server.begin();
}

void loop() {
  server.handleClient();
}




To σύνολο του κώδικα που χρησιμοποιούμε βρίσκεται εδώ:


#include <WiFi.h>
#include <WebServer.h>

// ================= PIN MAP =================
#define TILT_DIR   6
#define TILT_STEP  7
#define YAW_DIR    5
#define YAW_STEP   4
#define YAW_EN     20
#define TILT_EN    21

// ================= SCAN PARAMETERS =================

// Yaw: ~20° per move → 18 photos
const int yawStepsPerMove = 11;   // 1.8° × 11 ≈ 20°
const int yawPositions    = 18;

// Tilt: empirically chosen value
const int tiltSteps = 80;         // ⭐ good tilt for scanning

const int settleDelay    = 500;   // ms
const int cameraInterval = 2000;  // ms

WebServer server(80);

// ================= MOTOR CONTROL =================
void stepMotor(int stepPin, int dirPin, int steps, bool dir) {
  digitalWrite(YAW_EN, LOW);
  digitalWrite(TILT_EN, LOW);

  digitalWrite(dirPin, dir ? HIGH : LOW);

  for (int i = 0; i < steps; i++) {
    digitalWrite(stepPin, HIGH);
    delayMicroseconds(1200);
    digitalWrite(stepPin, LOW);
    delayMicroseconds(1200);
  }
}

void moveTiltUp() {
  stepMotor(TILT_STEP, TILT_DIR, tiltSteps, true);
}

void moveTiltDown() {
  stepMotor(TILT_STEP, TILT_DIR, tiltSteps, false);
}

// ================= SCAN =================
void scanOnce() {
  for (int i = 0; i < yawPositions; i++) {
    stepMotor(YAW_STEP, YAW_DIR, yawStepsPerMove, true);
    delay(settleDelay);
    delay(cameraInterval);
  }
}

void fullScan() {

  // --- Tilt center ---
  scanOnce();

  // --- Tilt UP ---
  moveTiltUp();
  delay(800);
  scanOnce();

  // --- Tilt DOWN ---
  moveTiltDown();
  moveTiltDown();   // go below center
  delay(800);
  scanOnce();

  // --- Back to center ---
  moveTiltUp();

  // Disable motors (cool down)
  digitalWrite(YAW_EN, HIGH);
  digitalWrite(TILT_EN, HIGH);
}

// ================= WEB UI =================
void handleRoot() {
  server.send(200, "text/html",
    "<h2>3D scanner – CDG, Cretan Dynamic Girls</h2>"

    "<p><b>Yaw:</b> 18 photos / full rotation</p>"
    "<p><b>Tilt:</b> ±" + String(tiltSteps) + " steps</p><hr>"

    "<button onclick=\"fetch('/start')\">Start Scan</button><br><br>"
    "<button onclick=\"fetch('/yaw')\">Yaw Step</button><br><br>"
    "<button onclick=\"fetch('/tiltUp')\">Tilt Up</button> "
    "<button onclick=\"fetch('/tiltDown')\">Tilt Down</button>"
  );
}

// ================= SETUP =================
void setup() {
  pinMode(TILT_DIR, OUTPUT);
  pinMode(TILT_STEP, OUTPUT);
  pinMode(YAW_DIR, OUTPUT);
  pinMode(YAW_STEP, OUTPUT);
  pinMode(YAW_EN, OUTPUT);
  pinMode(TILT_EN, OUTPUT);

  digitalWrite(YAW_EN, LOW);
  digitalWrite(TILT_EN, LOW);

  WiFi.softAP("3D-Scanner", "scanner123");

  server.on("/", handleRoot);

  server.on("/start", [](){
    fullScan();
    server.send(200, "text/plain", "Scan started");
  });

  server.on("/yaw", [](){
    stepMotor(YAW_STEP, YAW_DIR, yawStepsPerMove, true);
    server.send(200, "text/plain", "Yaw moved");
  });

  server.on("/tiltUp", [](){
    moveTiltUp();
    server.send(200, "text/plain", "Tilt up");
  });

  server.on("/tiltDown", [](){
    moveTiltDown();
    server.send(200, "text/plain", "Tilt down");
  });

  server.begin();
}

void loop() {
  server.handleClient();
}



## 🎉 Συγχαρητήρια!

Πλέον καταλαβαίνετε πώς ο κώδικας ελέγχει έναν πραγματικό μηχανισμό και πώς το λογισμικό συνδέεται με τη φυσική κίνηση.

Αυτό είναι **Ρομποτική + Προγραμματισμός + Φωτογραμμετρία** μαζί!
